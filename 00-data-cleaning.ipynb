```python
admission_df["los_hours"] = (
    admission_df["dischtime"] - admission_df["admittime"]
).dt.total_seconds() / 3600

admission_df = admission_df.merge(patients_df[["subject_id", "dob"]], on="subject_id")
admission_df.head()
```

```python
age_years = admission_df["admittime"].dt.year - admission_df["dob"].dt.year
birthday_not_yet = (
    (admission_df["admittime"].dt.month < admission_df["dob"].dt.month)
    | (
        (admission_df["admittime"].dt.month == admission_df["dob"].dt.month)
        & (admission_df["admittime"].dt.day < admission_df["dob"].dt.day)
    )
).astype("int64")
age_diff = age_years - birthday_not_yet

admission_df["age_at_admission"] = age_diff.apply(lambda x: 90 if x > 90 else x).astype(
    "int64"
)
```

```python
# admission_df['age_at_admission'] = ((admission_df['admittime'] - admission_df['dob']).dt.days / 365.25).astype('int64')
# admission_df['age_at_admission'] = ((admission_df['admittime'] - admission_df['dob']).dt.total_seconds()
# / (3600 * 24 * 365.25)).astype('int64')


admission_df = admission_df.sort_values(["subject_id", "admittime"])
admission_df["prev_dischtime"] = admission_df.groupby("subject_id")["dischtime"].shift(
    1
)
admission_df["days_since_last_discharge"] = (
    admission_df["admittime"] - admission_df["prev_dischtime"]
).dt.days
admission_df["readmission_30d"] = (
    admission_df["days_since_last_discharge"] <= 30
).astype(int)
```

```python
quality_log = []

future_dob = (
    pd.to_datetime(patients_df["dob"], errors="coerce") > pd.Timestamp.now()
).sum()
quality_log.append(
    {
        "Issue": "Dates shifted into future",
        "Table": "PATIENTS, ADMISSIONS",
        "Action": "Documented only — intentional anonymisation artefact. Age computed from date differences.",
        "Rows Affected": future_dob,
    }
)
print(
    f"Issue 1 — Future dates found in PATIENTS: {future_dob} rows (expected — anonymisation)"
)

print(quality_log)
```

```python
admission_df = admission_df.merge(
    patients_df[["subject_id", "dob"]], on="subject_id", how="left"
)

admission_df["los_hours"] = (
    admission_df["dischtime"] - admission_df["admittime"]
).dt.total_seconds() / 3600

invalid_los = admission_df["los_hours"] < 0
admission_df["timestamp_error"] = invalid_los.astype(int)

quality_log.append(
    {
        "Issue": "dischtime before admittime",
        "Table": "ADMISSIONS",
        "Action": "Flagged as 'timestamp_error' = 1. These rows excluded from LOS analysis.",
        "Rows Affected": invalid_los.sum(),
    }
)
print(
    f"Issue 2 — Invalid LOS (dischtime < admittime): {invalid_los.sum()} rows flagged"
)

print(quality_log)
```

```python
max_seq_diag = ICD_diagnoses.groupby("hadm_id")["seq_num"].max()
r_ICD_procedures = pd.read_csv(r"Raw-data/PROCEDURES_ICD.csv")
ICD_procedures = r_ICD_procedures.copy().drop(columns=["row_id"])
max_seq_proc = ICD_procedures.groupby("hadm_id")["seq_num"].max()

quality_log.append(
    {
        "Issue": "Multiple rows per hadm_id in diagnoses_icd / procedures_icd",
        "Table": "DIAGNOSES_ICD, PROCEDURES_ICD",
        "Action": "Expected behaviour — one row per ICD code. Logged max seq_num per admission.",
        "Rows Affected": f"Max seq_num diagnoses: {max_seq_diag.max()} | Max seq_num procedures: {max_seq_proc.max()}",
    }
)
print(f"Issue 3 — Max diagnosis codes per admission: {max_seq_diag.max()}")
print(f"Issue 3 — Max procedure codes per admission: {max_seq_proc.max()}")
```

```python
null_hadm_lab = Lab_events_df["hadm_id"].isnull().sum()
Lab_events_df["lab_link_status"] = Lab_events_df["hadm_id"].apply(
    lambda x: "unlinked_lab" if pd.isnull(x) else "linked"
)

quality_log.append(
    {
        "Issue": "NULL hadm_id in LABEVENTS",
        "Table": "LABEVENTS",
        "Action": "Flagged as 'unlinked_lab'. Rows kept in DB but excluded from admission-level joins.",
        "Rows Affected": null_hadm_lab,
    }
)
print(f"Issue 4 — Unlinked lab events (NULL hadm_id): {null_hadm_lab} rows")
```

```python
diag = ICD_diagnoses.merge(
    diagnoses_description_df[["icd9_code", "short_title"]], on="icd9_code", how="left"
)

diag["short_title"] = diag["short_title"].fillna("code_not_found")
unmatched_diag = (diag["short_title"] == "code_not_found").sum()

proc = ICD_procedures.merge(
    procedure_description_df[["icd9_code", "short_title"]], on="icd9_code", how="left"
)
proc["short_title"] = proc["short_title"].fillna("code_not_found")
unmatched_proc = (proc["short_title"] == "code_not_found").sum()

quality_log.append(
    {
        "Issue": "ICD-9 codes not found in lookup tables",
        "Table": "DIAGNOSES_ICD, PROCEDURES_ICD",
        "Action": "Left joined to lookups. Unmatched codes labelled 'code_not_found'.",
        "Rows Affected": f"Diagnoses: {unmatched_diag} | Procedures: {unmatched_proc}",
    }
)
print(
    f"Issue 5 — Unmatched ICD codes: Diagnoses={unmatched_diag}, Procedures={unmatched_proc}"
)
```

```python
Lab_events_df["outlier_flag"] = "normal"

lab_numeric = Lab_events_df[Lab_events_df["valuenum"].notnull()].copy()
stats = lab_numeric.groupby("itemid")["valuenum"].agg(["mean", "std"]).reset_index()
lab_numeric = lab_numeric.merge(stats, on="itemid", how="left")

outlier_mask = (
    abs(lab_numeric["valuenum"] - lab_numeric["mean"]) > 3 * lab_numeric["std"]
)
lab_numeric.loc[outlier_mask, "outlier_flag"] = "suspected_outlier"

Lab_events_df.update(lab_numeric[["outlier_flag"]])
outlier_count = (Lab_events_df["outlier_flag"] == "suspected_outlier").sum()

quality_log.append(
    {
        "Issue": "Extreme valuenum outliers in LABEVENTS",
        "Table": "LABEVENTS",
        "Action": "Values > 3 SD from mean per itemid flagged as 'suspected_outlier'.",
        "Rows Affected": outlier_count,
    }
)
print(f"Issue 6 — Suspected outliers in lab values: {outlier_count} rows")
```

```python
missing_discharge = admission_df["discharge_location"].isnull().sum()
admission_df["discharge_location"] = admission_df["discharge_location"].fillna(
    "UNKNOWN"
)

quality_log.append(
    {
        "Issue": "Missing discharge_location",
        "Table": "ADMISSIONS",
        "Action": "Imputed as 'UNKNOWN'.",
        "Rows Affected": missing_discharge,
    }
)
print(
    f"Issue 7 — Missing discharge_location: {missing_discharge} rows imputed as UNKNOWN"
)
```

```python
print("Before standardisation:", admission_df["admission_type"].unique())

admission_df["admission_type"] = admission_df["admission_type"].str.upper().str.strip()

# Map rare types to 'OTHER'
valid_types = ["EMERGENCY", "ELECTIVE", "URGENT", "NEWBORN"]
rare_before = (~admission_df["admission_type"].isin(valid_types)).sum()
admission_df["admission_type"] = admission_df["admission_type"].apply(
    lambda x: x if x in valid_types else "OTHER"
)

quality_log.append(
    {
        "Issue": "Inconsistent admission_type casing / rare types",
        "Table": "ADMISSIONS",
        "Action": "Standardised to UPPER CASE. Rare types mapped to 'OTHER'.",
        "Rows Affected": rare_before,
    }
)
print(f"Issue 8 — Rare/inconsistent admission types fixed: {rare_before} rows")
print("After standardisation:", admission_df["admission_type"].unique())
```

```python
adm = admission_df.sort_values(["subject_id", "admittime"])
adm["prev_dischtime"] = adm.groupby("subject_id")["dischtime"].shift(1)
adm["days_since_last_discharge"] = (adm["admittime"] - adm["prev_dischtime"]).dt.days
adm["readmission_30d"] = (adm["days_since_last_discharge"] <= 30).astype(int)

print(" \n Computed columns added successfully")
print(f"Avg LOS (hours): {adm[adm['timestamp_error'] == 0]['los_hours'].mean():.1f}")
print(f"Max age at admission: {adm['age_at_admission'].max()}")
print(f"30-day readmissions: {adm['readmission_30d'].sum()}")
```

```python
def clean_and_save(df, table_name_str, engine):
    df = df.drop_duplicates().copy()

    num_cols = df.select_dtypes(include="number").columns
    cat_cols = df.select_dtypes(include="object").columns

    for col in cat_cols:
        df[col] = df[col].astype(str).str.strip().str.title()
    for col in cat_cols:
        df[col] = df[col].fillna(df[col].mode()[0])

    for col in num_cols:
        if df[col].isnull().any():
            df[col] = df[col].fillna(df[col].mean())

        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

    df.to_sql(table_name_str, con=engine, if_exists="replace", index=False)

    print(f"Cleaned data saved to MySQL table: {table_name_str}")
    return df
```

```python
from sqlalchemy import create_engine, text

clean_and_save(diagnoses_description_df, "d_icd_diagnoses", engine)
clean_and_save(procedure_description_df, "d_icd_procedures", engine)
clean_and_save(patients_df, "patients", engine)
clean_and_save(admission_df, "admissions", engine)
clean_and_save(ICD_diagnoses, "diagnoses_icd", engine)
clean_and_save(ICD_procedures, "procedures_icd", engine)
clean_and_save(prescriptions_df, "prescriptions", engine)
clean_and_save(Lab_events_df, "labevents", engine)
```
